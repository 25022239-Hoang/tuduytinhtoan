Viết hàm kiểm tra một số có phải là số nguyên tố hay không?

```python

def snt(a):
    if a==1:
        return False
    else:
      for i in range(2,a):
          if a %i==0:
              return False
          else:
              return True
assert snt(10)==False
assert snt(3)==True
assert snt(11)==True
assert snt(22)==False
assert snt(7)==True

print('all test accepted')
```

VIết hàm kiểm tra một số có phải số hoàn hảo hay không, xử lý trên dữ liệu kiểu số nguyên

đầu tiên có giá trị k trong list số nguyên, nếu không có thì trả về -1 


```python
def lt(a,k):
    for i in range(len(a)):
        if a[i]==k:
            return i+1
        
    return -1
assert lt([1,12,13],12)==2
print('all test accepted')

         
```

Đầu ra: Kết quả phép toán

```python
def num(a,b,toantu):
    if toantu =='+':
        x=a+b
        return x
    elif toantu == '-':
        x=a-b
        return x
    elif toantu == '/':
        if b==0:
            return 'Error'
        else:
            x=a/b
            return f'{x:.2f}'
    elif toantu == '*':
        return (a*b)
    else:
        return 'Cút'
print(num(6,7,'*'))
```

Viết hàm compute(a, b, op) thực hiện phép tính a (+|-|*|/) b tùy theo ký tự op. 

```python
def compute(a, b, op):
    if op == "+":
        n=a+b
    elif op =="-":
        n=a-b
    elif op=="*":
        n=a*b
    else:
        if b==0:
            return False
        else :
            n=a/b
    return n
print(compute(6, 1, '/'))
```

Viết hàm kiểm tra một số có phải số Amstrong hay không? Hàm trả về kiểu bool, có tham số là một số
nguyên x


```python

def a(n):
    x=len(str(n))
    s=0
    z=n
    while n>0:
        y=n%10
        n=n//10
        s=s+y**x
    if s==z:
        return True
    return False
assert a(153)==True
assert a(154)==False
print('all test accepted')
```

```python
import math
def scp(n):
    can=int(math.sqrt(n))
    if can*can==n:
        return True
    else:
        return False
assert scp(64)==True
assert scp(1)==True
assert scp(2)==False
print('all test accepted')
    
```

W5A18 VIết hàm kiểm tra một số có phải số đối xứng không, xử lý trên dữ liệu kiểu số nguyên.

```python
def sdx(n):
    a=int(str(n)[::-1])
    if a==n:
        return True
    else:
        return False
assert sdx(67)==False
assert sdx(65)==False
assert sdx(121)==True
assert sdx(131)==True
assert sdx(112)==False
print('all test accepted')
```
